```python
import os
from dotenv import load_dotenv
from agents import Agent, Runner, trace, AsyncOpenAI, OpenAIChatCompletionsModel  # Add this import
import asyncio

```

```python
# The usual starting point

load_dotenv(override=True)

print("Grok Base URL:", os.getenv("GROK_BASE_URL"))
print("Grok API Key:", os.getenv("GROK_API_KEY")[0:4] + "****")

# create the Grok client for use in the agent
grok_client = AsyncOpenAI(
    base_url=os.getenv("GROK_BASE_URL"),  # Should be "https://api.x.ai/v1" or similar
    api_key=os.getenv("GROK_API_KEY")
)
```

```python

# Make an agent with name, instructions, model

# agent = Agent(name="Jokester", instructions="You are a joke teller", model="gpt-4o-mini")
```

```python
grok_client = AsyncOpenAI(
    base_url=os.getenv("GROK_URL"),  # Or hardcode if needed, but don't commit it
    api_key=os.getenv("GROK_API_KEY")  # Or hardcode if needed, but don't commit it
)

grok_agent = Agent(
    name="Jokester",
    instructions="You are a joke teller",
    model=OpenAIChatCompletionsModel(
        model="grok-4",  # Replace with the exact Grok model name from xAI docs
        openai_client=grok_client
    )
)
```

```python
# Run the joke with Runner.run(agent, prompt) then print final_output

# with trace("Telling a joke"):
#     result = await Runner.run(agent, "Tell a joke about Autonomous AI Agents")
#     print(result.final_output)
```

```python
with trace("Telling a joke"):
    result = await Runner.run(grok_agent, input="Tell me a joke about AI.")
    print(result.final_output)
```
