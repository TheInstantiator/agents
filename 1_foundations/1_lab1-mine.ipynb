```python
# Check the key - if you're not using OpenAI, check whichever key you're using! Ollama doesn't need a key.

import os
openai_api_key = os.getenv('OPENAI_API_KEY')
grok_api_key = os.getenv('GROK_API_KEY')
gemini_api_key = os.getenv('GEMINI_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set - please head to the troubleshooting guide in the setup folder")

if grok_api_key:
    print(f"Grok API Key exists and begins {grok_api_key[:8]}")
else:
    print("Grok API Key not set - please head to the troubleshooting guide in the setup folder")

if gemini_api_key:
    print(f"Gemini API Key exists and begins {gemini_api_key[:8]}")
else:
    print("Gemini API Key not set - please head to the troubleshooting guide in the setup folder")
```

```python
# OpenAI is a class in the openai module

from openai import OpenAI
```

```python
# create an instance of the OpenAI class
# this instance will be used to make API calls
# openai ia a client object
openai = OpenAI()

# this seems to be free so I won't comment it out, I don't have any billing set up on this key or for openai
try:
  models = openai.models.list()
  all_models = [m.id for m in models.data]
  print("OpenAI models available:", all_models)
except Exception as e:
  print("Error accessing OpenAI models:", e)
```

```python
# create a Grok client

GROK_BASE_URL = "https://api.x.ai/v1"
grok = OpenAI(base_url=GROK_BASE_URL, api_key=grok_api_key)

try:
    models = grok.models.list()
    # Print all model IDs (raw output)
    all_models = [m.id for m in models.data]
    print("All model IDs:", all_models)
except Exception as e:
    print(f"Error fetching Grok models: {e}")
```

```python
# create a Gemini client

GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=gemini_api_key)

try:
    models = gemini.models.list()
    # Print all model IDs (raw output)
    all_models = [m.id for m in models.data]
    print("All Gemini model IDs:", all_models)
except Exception as e:
    print(f"Error fetching Gemini models: {e}")
```

```python
# Create a list of messages in the familiar OpenAI format

messages = [{"role": "user", "content": "What is 2+2?"}]
```

```python
# First create the messages:

messages = [{"role": "user", "content": "propose a business area worth exploring in the agentic ai space"}]
```

```python
# Then make the first call:

grok_response = grok.chat.completions.create(
    model="grok-4",
    messages=messages
)

gemini_response = gemini.chat.completions.create(
    model="gemini-2.5-pro",
    messages=messages
)

# Then read the business idea:

grok_business_idea = grok_response.choices[0].message.content
gemini_business_idea = gemini_response.choices[0].message.content
print("GROK: ", grok_business_idea)
print("GEMINI: ", gemini_business_idea)
```

```python
messages = [{"role": "user", "content": f"Propose a pain point in this area worth exploring for creating an agentic solution in the agentic AI space. Here is an idea to build on: {grok_business_idea}"}]

grok_response = grok.chat.completions.create(
    model="grok-4",
    messages=messages
)

grok_pain_point = grok_response.choices[0].message.content
print(grok_pain_point)
```

```python
messages = [{"role": "user", "content": f"propose a pain point in this area worth exploring creating an agentic solution for in the agentic ai space. Here is an idea to build on: {gemini_business_idea}"}]

gemini_response = gemini.chat.completions.create(
    model="gemini-2.5-pro",
    messages=messages
)
gemini_pain_point = gemini_response.choices[0].message.content
print(gemini_pain_point)
```

```python
messages = [{"role": "user", "content": f"propose an agentic ai solution to solve this pain point: {grok_pain_point}"}]

grok_response = grok.chat.completions.create(
    model="grok-4-fast-reasoning",
    messages=messages
) 
print(grok_response.choices[0].message.content)
```

```python
messages = [{"role": "user", "content": f"propose an agentic ai solution to solve this pain point: {gemini_pain_point}"}] 

gemini_response = gemini.chat.completions.create(
    model="gemini-2.5-pro",
    messages=messages
)
print(gemini_response.choices[0].message.content)
```
